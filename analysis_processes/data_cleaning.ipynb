```python
import pandas as pd
import openpyxl
```

```python
""" READ ALL SHEET """
all_sheet = pd.read_excel(file_path, sheet_name=None)

print(f"List sheet: {list(all_sheet.keys())}")
```

```python
""" ACCESS EVERY SHEET """
df_sales_orders = all_sheet["Sales_Orders"]
df_purchase_orders = all_sheet["Purchase_Orders"]
df_product_master = all_sheet["Product_Master"]
df_supplier_master = all_sheet["Supplier_Master"]
df_warehouse_master = all_sheet["Warehouse_Master"]
```

**FUNCTION LIST**

```python
# For Checking Date String
def date_data_string (data) :
  return data[data.apply(lambda x: isinstance(x, str))]
```

```python
# Convert Date String to Datetime
def convert_to_datetime (data) :
  return pd.to_datetime (
      data,
      format="mixed",
      errors="coerce"
  )
```

```python
# Dropna function
def drop_na (data, subset_value):
  return data.dropna(subset=[subset_value])
```

```python
# Check Unique Data
def check_unique_data (data):
  return data.unique()
```

```python
# Cleaning SKU
def cleaning_SKU (data):
  return "SKU-" + data.astype(str).str.strip().str[-4:]
```

```python
# Check Excess Spaces
def check_excess_spaces (data):
  return data[data != data.astype(str).str.strip()]
```

```python
# Not match data with their title()
def check_not_match_title (data):
  return data[data != data.astype(str).str.title()]
```

```python
# Delete Extra Spaces
def delete_extra_spaces_string (data) :
  return data.astype(str).str.strip().str.replace(f'\s+', ' ', regex=True)
```

```python
# Convert data to numeric
def numeric_converter (data):
  return pd.to_numeric (
      data,
      errors="coerce"
  )
```

```python
# Check Numeric Na Value
def check_numeric_na (raw_data, converted_data):
  return raw_data[raw_data.notna() & converted_data.isna()]
```

```python
# Check Negative Value
def check_negative_value (raw_data, converted_data):
  return raw_data[converted_data < 0]
```

```python
def check_column_issue (raw_data):
  converted_data = numeric_converter(raw_data)
  failed_converted_data = check_numeric_na(raw_data, converted_data)
  negative_values = check_negative_value (raw_data, converted_data)

  print(f"Failed Converted Data:\n{list(failed_converted_data)}")
  print(f"\nNegative Values:\n{list(negative_values)}")
```

```python
# Clean Additional Characters
def clean_characters (raw_data):
  return pd.to_numeric (
      raw_data.astype(str)
      .str.strip()
      .str.replace(r'[^\d]', '', regex=True)
      ,errors="coerce"
  ).abs()
```

```python
def check_duplicates(raw_data):
    dup_mask = raw_data.duplicated()

    duplicate_total = dup_mask.sum()
    duplicate_rows = raw_data[dup_mask]
    duplicate_indexes = duplicate_rows.index

    print("=== ANALYSIS RESULT ====")
    print(f"Duplicate Total   : {duplicate_total}")
    print(f"Duplicate Indexes : {duplicate_indexes.tolist()}")
    print(f"Duplicate Rows    : \n{duplicate_rows}\n")

    return {
        "total" : duplicate_total,
        "indexes" : duplicate_indexes,
        "rows" : duplicate_rows
    }
```

**PRODUCT MASTER SECTION**

2. Category

```python
# Check Data Variation
raw_data = df_product_master["Category"]

title_check = check_not_match_title(raw_data)
print(f"Title Check:\n{list(title_check)}")

unique_data = check_unique_data(raw_data)
print(f"Unique Data:\n{list(unique_data)}")

# We have 'RAW MATERIAL'
df_product_master["Category"] = df_product_master["Category"].astype(str).str.title()
```

```python
df_sales_orders.loc[df_sales_orders["Order_Status"] == "Cancelled", "Quantity_Sold"] = 0

median_qty = df_sales_orders.groupby("SKU")["Quantity_Sold"].transform("median")
df_sales_orders["Quantity_Sold"] = df_sales_orders["Quantity_Sold"].fillna(median_qty)
```

**Fixed Purchase Orders**

```python
df_purchase_orders["Product_Name"] = delete_extra_spaces_string(df_purchase_orders["Product_Name"]).str.title()

df_purchase_orders = df_purchase_orders.drop(columns=["SKU"])

df_purchase_orders = pd.merge(
    df_purchase_orders,
    df_product_master[["Product_Name", "SKU"]].drop_duplicates("Product_Name"),
    on="Product_Name",
    how="inner"
)
```

```python
df_purchase_orders = df_purchase_orders[
    ["PO_Number","PO_Date","SKU","Product_Name","Supplier_ID","Warehouse_ID","Quantity_Ordered","Unit_Cost","Required_Date","Actual_Delivery_Date","PO_Status",]
]
```

```python
# Fill NA
df_purchase_orders = pd.merge(
    df_purchase_orders,
    df_product_master[["SKU", "Lead_Time_Days"]],
    on="SKU",
    how="left"
)

estimated_required_date = df_purchase_orders["PO_Date"] + pd.to_timedelta(df_purchase_orders["Lead_Time_Days"], unit="D")
df_purchase_orders["Required_Date"] = df_purchase_orders["Required_Date"].fillna(estimated_required_date)

df_purchase_orders = df_purchase_orders.drop(columns=["Lead_Time_Days"])

received_mask = (df_purchase_orders["PO_Status"] == "Received") & (df_purchase_orders["Actual_Delivery_Date"].isna())
df_purchase_orders.loc[received_mask, "Actual_Delivery_Date"] = df_purchase_orders.loc[received_mask, "Required_Date"]
```

```python
df_purchase_orders = df_purchase_orders.dropna(subset=["Quantity_Ordered"]).reset_index(drop=True)
```

```python
clean_dataframes = {
    "sales_orders_clean": df_sales_orders,
    "purchase_orders_clean": df_purchase_orders,
    "product_master_clean": df_product_master,
    "supplier_master_clean": df_supplier_master,
    "warehouse_master_clean": df_warehouse_master,
}

output_path = '/Clean_Data/'

for name, df in clean_dataframes.items():
  df.to_csv(f"{output_path}{name}.csv", index=False)
```
